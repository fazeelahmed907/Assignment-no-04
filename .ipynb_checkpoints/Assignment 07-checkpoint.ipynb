```python
class Employee:
    def __init__(self, name, number, address, salary, post, joining_year, education, experience):
        self.name = name
        self.number = number
        self.address = address
        self.salary = salary
        self.post = post
        self.joining_year = joining_year
        self.education = education
        self.experience = experience

    def describe_employee(self):
        print(self.name, self.number, self.address, self.salary,
              self.post, self.joining_year, self.education, self.experience)

    def update_year_of_experience(self, years):
        self.experience += years

    def increase_salary(self, amount):
        self.salary += amount

```

```python
employee101 = Employee("Ali", 101, "Karachi", 50000, "Manager", 2020, "MBA", 5)

print(employee101.name)
print(employee101.number)
print(employee101.address)
print(employee101.salary)

employee101.describe_employee()
employee101.increase_salary(5000)
employee101.update_year_of_experience(1)
employee101.describe_employee()

employees = [
    Employee("Ahmed", 102, "Lahore", 40000, "Clerk", 2021, "B.Com", 3),
    Employee("Sara", 103, "Karachi", 50000, "Teacher", 2020, "MA", 4),
    Employee("Usman", 104, "Multan", 60000, "Manager", 2019, "MBA", 6),
    Employee("Ayesha", 105, "Islamabad", 45000, "Accountant", 2022, "BBA", 2),
    Employee("Hassan", 106, "Quetta", 70000, "Developer", 2018, "BSCS", 7),
    Employee("Fatima", 107, "Lahore", 55000, "HR", 2021, "MBA", 3),
    Employee("Bilal", 108, "Karachi", 35000, "Assistant", 2023, "BA", 1),
    Employee("Zain", 109, "Peshawar", 65000, "Designer", 2019, "BSCS", 5),
    Employee("Hina", 110, "Multan", 48000, "Supervisor", 2020, "BBA", 4),
    Employee("Hamza", 111, "Sialkot", 80000, "Director", 2017, "MBA", 8)
]

for emp in employees:
    emp.describe_employee()
    emp.increase_salary(5000)
    emp.update_year_of_experience(1)

for emp in employees:
    emp.describe_employee()
```
