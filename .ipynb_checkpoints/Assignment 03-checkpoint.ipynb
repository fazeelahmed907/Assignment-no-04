```python

age = int(input("Enter your age: "))

if age < 0:
    print("Invalid age")
elif age <= 12:
    print("Child")
elif age <= 19:
    print("Teenager")
elif age <= 64:
    print("Adult")
else:
    print("Senior")
```

```python
course_a = {"John", "Alice", "Bob", "David"}
course_b = {"Alice", "Eve", "Charlie", "David"}
# Difference: Students in Course A but not in Course B
difference = course_a.difference(course_b)
difference
```

```python
# Symmetric Difference: Students in only one course
symmetric_difference = course_a.symmetric_difference(course_b)
symmetric_difference
```

```python
A = {"Ali", "Sara"}
B = {"Ahmed", "Zara"}
#Combine both sets into one set.
A.union(B)
```

```python
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7}

#Find common elements.Find elements only in Set A.Find elements only in Set B.
A.intersection(B)
```

```python
#Find elements only in Set A.
A.difference(B)
```

```python
#Find elements only in Set B.
B.difference(A)
```
