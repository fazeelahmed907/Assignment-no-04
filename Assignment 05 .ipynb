```python
def count_vowels(vowels):
    count = 0

    for count_vowels in vowels:
        if count_vowels in "aeiouAEIOU":
            count += 1

    return count

vowels = "Welcome"
print(count_vowels(vowels), "vowels")
```

```python
delivery_charges = lambda order: 150 if order < 2000 else 0

print("Delivery Charges:", delivery_charges(1500))
print("Delivery Charges:", delivery_charges(2500))
```

```python
def calculate_rectangle(length, width):
    def calculate_area():
        return length * width

    def calculate_perimeter():
        return 2 * (length + width)

    return calculate_area(), calculate_perimeter()

area, perimeter = calculate_rectangle(5, 3)

print("Area:", area)
print("Perimeter:", perimeter)
```

```python
def hotel_bill(nights):
    def calculate_tax(cost):
        return cost * 0.10

    room_cost = nights * 5000
    tax = calculate_tax(room_cost)
    total = room_cost + tax

    print("Room Cost:", room_cost)
    print("Tax:", tax)
    print("Total Bill:", total)

hotel_bill(3)
```

```python
def login(username, password):
    def validate():
        return username == "admin" and password == "1234"

    if validate():
        print("Login successful!")
    else:
        print("Invalid username or password.")

login("admin", "1234")
```

```python
def shopping_bill(price, quantity, discount=0):

    def calculate_total():
        return price * quantity

    def apply_discount(total):
        return max(0, total - discount)

    return apply_discount(calculate_total())


print("Final Bill =", shopping_bill(200, 3), "PKR")
print("Final Bill =", shopping_bill(200, 3, 100), "PKR")
```
