```python
products = {
    "P1": {"name": "Shoes", "price": 2500, "stock": 5},
    "P2": {"name": "Shirt", "price": 1500, "stock": 10},
    "P3": {"name": "Bag", "price": 3000, "stock": 4},
    "P4": {"name": "Watch", "price": 4000, "stock": 3},
    "P5": {"name": "Headphones", "price": 2000, "stock": 6}
}

cart = {}

while True:

    print("\n===== ONLINE SHOPPING CART =====")
    print("1. View Products")
    print("2. Add to Cart")
    print("3. View Cart")
    print("4. Checkout")
    print("5. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        print("\n--------- AVAILABLE PRODUCTS ---------")
        print("ID\tName\t\tPrice\t\tStock")

        for pid, details in products.items():
            print(f"{pid}\t{details['name']}\t\tRs.{details['price']}\t\t{details['stock']}")

    elif choice == "2":
        pid = input("Enter Product ID: ").upper()

        if pid in products:
            quantity = int(input("Enter Quantity: "))

            if quantity <= 0:
                print("Please enter a valid quantity.")

            else:
                cart_quantity = cart.get(pid, 0) + quantity

                if cart_quantity <= products[pid]["stock"]:
                    cart[pid] = cart_quantity
                    print("Product added to cart successfully!")

                else:
                    print("Sorry! Not enough stock available.")

        else:
            print("Invalid Product ID!")

    elif choice == "3":
        if not cart:
            print("\nYour cart is empty.")

        else:
            total = 0

            print("\n--------- YOUR SHOPPING CART ---------")

            for pid, quantity in cart.items():
                name = products[pid]["name"]
                price = products[pid]["price"]
                subtotal = price * quantity
                total += subtotal

                print(f"{name} | Quantity: {quantity} | "
                      f"Price: Rs.{price} | Subtotal: Rs.{subtotal}")

            print("--------------------------------------")
            print("Cart Total: Rs.", total)

    elif choice == "4":
        if not cart:
            print("\nYour cart is empty. Add products first.")

        else:
            total = 0

            print("\n========== FINAL BILL ==========")

            for pid, quantity in cart.items():
                name = products[pid]["name"]
                price = products[pid]["price"]
                subtotal = price * quantity
                total += subtotal

                print(f"{name} x {quantity} = Rs.{subtotal}")

            print("--------------------------------")

            if total >= 5000:
                discount = total * 0.10
            else:
                discount = 0

            final_bill = total - discount

            print("Total Amount: Rs.", total)
            print("Discount: Rs.", discount)
            print("Final Bill: Rs.", final_bill)

            for pid, quantity in cart.items():
                products[pid]["stock"] -= quantity

            cart.clear()

            print("\nCheckout Successful!")
            print("Thank you for shopping with us.")

    elif choice == "5":
        print("\nThank you for visiting our store!")
        break

    else:
        print("Invalid choice! Please try again.")
```
